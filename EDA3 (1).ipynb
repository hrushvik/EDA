```python
import numpy as np
arr=np.arange(0,21,1)
print(arr)
print(arr.ndim)
print(arr.shape)
print(arr.size)
print(arr.dtype)
```

```python
np.random.seed(1)
arr=np.random.randint(10,size=(4,5))
print(arr)
print(arr.size)
print(arr.nbytes)
```

```python
np.random.seed(1)
arr=np.arange(1,13).reshape(3,4)
print(arr)
print(arr[1,])
print(arr[:,2])
print(arr[-1,])
print(arr[2,])
```

```python
np.random.seed(1)
arr=np.random.randint(10,size=(3,4))
print(arr)
print(arr[:2,:2])
```
