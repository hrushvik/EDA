```python
import numpy as np
import pandas as pd

data=pd.Series([1,2,3,4,5])
print(data)
data=pd.Series([1,2,3,4,5],index=['a','b','c','d','e'])
print(data)
print(data.values)
print(data.index)
```

```python
data=pd.Series({2:'a',3:'b',4:'e'})
print(data)
```

```python
df=pd.DataFrame({'1st':data,'2nd':data})
print(df)
```

```python
arr=np.array([[1,2,3],[44,55,66]])
dataframe=pd.DataFrame(arr)
print(dataframe)
```

```python
ind=pd.Index([2,3,5,7,11])
print(ind.size,ind.shape,ind.ndim,ind.dtype)
print(ind[1])
print(ind[::2])
print(ind[1:4:2])
indA=pd.Index([1,2,3,4,5])
indB=pd.Index([6,7,8,9,0])
print(indA & indB)
print(indA | indB)
print(indA ^ indB)
print(indA.intersection(indB))
```
